```python
import pandas as pd
```

```python
data = {
    'Nama': ['Alice', 'Bob', 'Charlie'],
    'Umur': [25, 30, 35],
    'Kota': ['New York', 'Los Angeles', 'Chicago']
}
df = pd.DataFrame(data)
print(df)
```

```python
# Reading File
# df = pd.read_csv('data.csv')
print(df.head())  # Menampilkan lima baris pertama
```

```python
df.set_index('Nama', inplace=True)
print(df)
```

```python
filtered_df = df[df['Umur'] > 30]
print(filtered_df)
```
